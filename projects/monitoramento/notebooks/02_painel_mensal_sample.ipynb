```python
from pathlib import Path
import sys, os
NB_DIR = Path.cwd()
MONO = next((p for p in [NB_DIR, *NB_DIR.parents] if (p / "shared" / "cemepi_api").exists()), NB_DIR)
sys.path.insert(0, str(MONO))
os.chdir(MONO)
from shared.cemepi_api import CemepiClient, load_settings
S = load_settings(); C = CemepiClient(S)
print("extracao", S.ano, S.mes, "limit", S.sample_limit)
```

```python
import pandas as pd
def first_list(o):
  if isinstance(o,list): return o
  if isinstance(o,dict):
    for v in o.values():
      if isinstance(v,list) and v and isinstance(v[0],dict): return v
  return []
e=pd.DataFrame(first_list(C.get_json("/v1/debito/estoque"))); a=pd.DataFrame(first_list(C.get_json("/v1/arrecadacao/serie")))
for df in (e,a): df.rename(columns={c:c.lower() for c in df.columns},inplace=True)
keys=[k for k in ["ano","mes"] if k in e.columns and k in a.columns]
panel=e.merge(a,on=keys,how="inner",suffixes=("_estoque","_arrec")) if keys else e
panel.to_parquet(C.sample_dir()/"b2_painel_mensal_sample.parquet",index=False)
print(keys, panel.shape)
```
