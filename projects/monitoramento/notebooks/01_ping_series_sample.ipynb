```python
import pandas as pd
def first_list(o):
  if isinstance(o,list): return o
  if isinstance(o,dict):
    for v in o.values():
      if isinstance(v,list) and v and isinstance(v[0],dict): return v
  return []
e=pd.DataFrame(first_list(C.get_json("/v1/debito/estoque")))
a=pd.DataFrame(first_list(C.get_json("/v1/arrecadacao/serie")))
e.head(120).to_parquet(C.sample_dir()/"b1_estoque_head.parquet",index=False)
a.head(120).to_parquet(C.sample_dir()/"b1_arrec_head.parquet",index=False)
print(e.shape,a.shape)
```
